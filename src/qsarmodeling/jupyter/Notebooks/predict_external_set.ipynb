```python
import numpy as np
import pandas as pd

from qsarmodelingpy.lj_cut import transform
from qsarmodelingpy.cross_validation_class import CrossValidation

from sklearn.cross_decomposition import PLSRegression
```

```python
directoryExt = "/home/jpam/MEGA/eduardo/DAT/triagem/LQTAGridFiles/"
directoryTrain = "/home/jpam/MEGA/eduardo/DAT/LQTAGridFiles/"
```

```python
dfX = pd.read_csv(directoryTrain + "Xtrain.csv", sep=";", index_col=0)
y = pd.read_csv(directoryTrain + "ytrain.csv", sep=",", header=None, index_col=0).values
X = dfX.values
```

```python
dfXext = pd.read_csv(directoryExt + "matriz_previsao4D.csv", sep=";", index_col=0)
```

```python
desc = dfX.columns
```

```python
dfExt = transform(dfXext)
```

```python
dfXextSel = dfExt.loc[:, desc]
```
