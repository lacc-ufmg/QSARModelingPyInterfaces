```python
# Mount the prediction matrix using the selected variables in the calibration step
import pandas as pd
directory = "/home/jpam/MEGA/eduardo/DAT/3D/LQTAGridFiles/"
calibration_matrix_file = directory + "Xtrain.csv"
dependent_variable_file = directory + "ytrain.csv"
prediction_matrix_file = "/home/jpam/MEGA/eduardo/DAT/triagem/3D/matriz_previsao.csv"
calibration_matrix = pd.read_csv(calibration_matrix_file,sep=';',index_col=0)
prediction_matrix = pd.read_csv(prediction_matrix_file,sep=';',index_col=0)
dependent_variable = pd.read_csv(dependent_variable_file,sep=',',index_col=0,header=None)
var_sel = calibration_matrix.columns
prediction_selected_matrix = prediction_matrix.loc[:,var_sel]
prediction_selected_matrix.to_csv("/home/jpam/MEGA/eduardo/DAT/teste_ext_val/Xpred.csv",sep=';')
```

```python
# Prediction of the real external set
from sklearn.cross_decomposition import PLSRegression
# Fill the number of latent variables in the model
nLV = 2
# Transform LJ values. Run the following three lines only if you want to preform this transformation
from lj_cut import transform
X = transform(calibration_matrix)
Xp = transform(prediction_selected_matrix)
pls = PLSRegression(n_components = nLV)
pls.fit(X,dependent_variable)
yp = pls.predict(Xp)
dfPrediction = pd.DataFrame(yp)
dfPrediction.index = prediction_matrix.index
dfPrediction
```
